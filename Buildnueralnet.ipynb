```python
learning_rate = 0.1

w = w - learning_rate * gradient

print("New Weight:", w)
```

```python
import numpy as np

# Dataset
X = np.array([1, 2, 3, 4, 5])
Y = np.array([0, 0, 0, 1, 1])

# Parameters
w = 0.5
b = 0.0

learning_rate = 0.1

epochs = 10
```

```python
for epoch in range(epochs):

    total_loss = 0

    for x, y in zip(X, Y):

        # Forward
        z = w * x + b
        a = 1 / (1 + np.exp(-z))

        # Loss
        loss = 0.5 * (y - a) ** 2

        total_loss += loss

        # Backward
        dL_da = a - y
        da_dz = a * (1 - a)
        dz_dw = x
        dz_db = 1

        gradient_w = dL_da * da_dz * dz_dw
        gradient_b = dL_da * da_dz * dz_db

        # Update
        w -= learning_rate * gradient_w
        b -= learning_rate * gradient_b

    print(
        f"Epoch {epoch+1} "
        f"Loss={total_loss:.4f} "
        f"Weight={w:.4f} "
        f"Bias={b:.4f}"
    )
```
